```python
import pandas as pd
from src.data.loader import load_dataset, inspect_dataframe, print_inspection, STANDARD_COLUMNS
pd.set_option("display.max_columns", 50)

# EDIT these paths to where the datasets live on your machine / Drive.
RAW_PATHS = {
    "fakehealth": "data/raw/FakeHealth",
    "coaid":      "data/raw/CoAID",
    "pubhealth":  "data/raw/PUBHEALTH",
}
```

## Load each dataset (datasets whose folder is missing are skipped, not faked)

```python
datasets, summaries = {}, {}
for name, path in RAW_PATHS.items():
    if not Path(path).exists():
        print(f"[skip] {name}: {path} not found"); continue
    try:
        datasets[name] = load_dataset(name, path)
        summaries[name] = inspect_dataframe(datasets[name], name)
    except Exception as e:
        print(f"[error] {name}: {e}")
print("Loaded:", list(datasets))
```

## Inspection (take screenshots of these outputs for the report)

```python
for name, df in datasets.items():
    print_inspection(summaries[name])
    display(df.head(3))
    print()
```

## Save raw standardized snapshots (input for notebook 02)

```python
Path("data/processed").mkdir(parents=True, exist_ok=True)
for name, df in datasets.items():
    df.to_csv(f"data/processed/{name}_raw_standardized.csv", index=False)
    print("saved", name, df.shape)
```
